
# 1. VERİ SETİNİN TANITILMASI VE GENEL İNCELENMESİ

## 1.1. Çalışmanın Konusu ve Amacı

**Proje Başlığı:** Veri Madenciliği Teknikleri ile IoT/IIoT Ağlarında Siber Saldırı Tespiti ve Sınıflandırılması

**Çalışma Alanı:** Siber Güvenlik, Nesnelerin İnterneti (IoT), Endüstriyel Nesnelerin İnterneti (IIoT) ve Veri Madenciliği

### Giriş

Nesnelerin İnterneti (IoT) ve Endüstriyel Nesnelerin İnterneti (IIoT) teknolojilerinin yaygınlaşması, ağlara bağlanan cihaz sayısının artmasına ve bu cihazlar arasında büyük miktarda veri trafiği oluşmasına neden olmaktadır. Bununla birlikte ağ altyapılarında meydana gelen siber saldırıların tespit edilmesi, bilgi güvenliği açısından önemli bir gereksinim hâline gelmiştir.

Geleneksel güvenlik yöntemlerinin yanı sıra veri madenciliği ve makine öğrenmesi teknikleri, ağ trafiği içerisindeki olağan ve olağan dışı davranışların analiz edilmesinde kullanılmaktadır. Özellikle ağ trafiğinin çeşitli öznitelikler üzerinden incelenmesi, normal iletişim davranışları ile saldırı faaliyetlerinin birbirinden ayrılmasına katkı sağlamaktadır.

Bu proje kapsamında, IoT ve IIoT ortamlarındaki ağ trafiğini temsil eden **Edge-IIoTset (2022)** veri seti kullanılmaktadır. Çalışmanın temel amacı, veri madenciliği yöntemleriyle ağ kayıtlarını analiz etmek, normal trafik ile saldırı trafiğini ayırt etmek ve farklı siber saldırı türlerini sınıflandırabilen makine öğrenmesi modelleri geliştirmektir.

### 1.2. Edge-IIoTset Veri Setinin Genel Özellikleri

Edge-IIoTset, IoT ve IIoT sistemlerinde normal ağ faaliyetleri ile çeşitli siber saldırı davranışlarını içeren, siber güvenlik araştırmalarında kullanılabilen bir veri setidir.

Veri seti; ağ iletişimi, protokol özellikleri ve trafik davranışlarına ilişkin çeşitli bilgileri içermektedir. TCP, UDP, HTTP, DNS, MQTT ve diğer ağ protokollerine ait öznitelikler, trafik kayıtlarının analiz edilmesinde kullanılabilecek bilgiler sağlamaktadır.

Çalışma kapsamında incelenen veri seti klasöründe toplam **26 CSV dosyası** bulunmaktadır:

- **14 CSV dosyası:** Siber saldırı trafiğine ilişkin kayıtlar.
- **10 CSV dosyası:** Normal trafik ve sensör verilerine ilişkin kayıtlar.
- **2 CSV dosyası:** Makine öğrenmesi ve derin öğrenme çalışmaları için hazırlanmış veri dosyaları.

Makine öğrenmesi için hazırlanmış `ML-EdgeIIoT-dataset.csv` dosyası 157.800 kayıt ve 63 sütundan oluşmaktadır. Bu sütunların 61'i öznitelik, 2'si ise hedef değişkendir.

Bu çalışmada, veri temizleme, örnekleme ve öznitelik hazırlama işlemlerinin doğrudan incelenebilmesi amacıyla **24 ham CSV dosyası** temel alınmıştır.

### 1.3. Hedef Değişkenlerin Tanımlanması

Veri setinde iki temel hedef değişken bulunmaktadır:

**Attack_label:** Ağ trafiğinin normal veya saldırı durumunu temsil eden ikili hedef değişkendir.

- `0`: Normal trafik
- `1`: Saldırı trafiği

**Attack_type:** Ağ trafiğinin ait olduğu sınıfı göstermektedir. Bu değişken, normal trafik ve farklı siber saldırı türlerinin çok sınıflı sınıflandırılmasında kullanılmaktadır.

Bu iki değişken sayesinde çalışmada hem ikili saldırı tespiti hem de saldırı türü sınıflandırması değerlendirilebilecektir.

```python
import pandas as pd

# ML veri setinin dosya yolunu belirleme
dosya_yolu = next(
    veri_klasoru.rglob("ML-EdgeIIoT-dataset.csv")
)

# Bellek kullanımını azaltmak için verileri parçalar halinde okuma
toplam_kayit = 0
etiket_sayilari = pd.Series(dtype="int64")
saldiri_sayilari = pd.Series(dtype="int64")

for parca in pd.read_csv(
    dosya_yolu,
    usecols=["Attack_label", "Attack_type"],
    chunksize=10000
):
    toplam_kayit += len(parca)

    etiket_sayilari = etiket_sayilari.add(
        parca["Attack_label"].value_counts(),
        fill_value=0
    )

    saldiri_sayilari = saldiri_sayilari.add(
        parca["Attack_type"].value_counts(),
        fill_value=0
    )

# Sonuçları gösterme
print("Toplam kayıt sayısı:", toplam_kayit)

print("\nNormal ve saldırı kayıtlarının dağılımı:")
print(etiket_sayilari.astype(int))

print("\nSaldırı türlerine göre kayıt sayıları:")
print(saldiri_sayilari.astype(int).sort_values(ascending=False))
```

```python

import pandas as pd
from pathlib import Path

# Veri setinin bulunduğu klasörü belirleme
veri_klasoru = Path(
    r"E:\Data mining\Network_Intrusion_Detection\archive\Edge-IIoTset dataset"
)

# ML veri setinin dosyasını bulma
dosya_yolu = next(veri_klasoru.rglob("ML-EdgeIIoT-dataset.csv"))

# Analiz sonuçları için değişkenler
toplam_kayit = 0
eksik_satir_sayisi = 0
eksik_sayilari = None

# Bellek kullanımını azaltmak için veriyi parçalar halinde okuma
for parca in pd.read_csv(dosya_yolu, chunksize=10000):

    toplam_kayit += len(parca)

    # Eksik değerleri sütun bazında sayma
    parca_eksik = parca.isna().sum()

    if eksik_sayilari is None:
        eksik_sayilari = parca_eksik
    else:
        eksik_sayilari = eksik_sayilari + parca_eksik

    # En az bir eksik değer içeren satırları sayma
    eksik_satir_sayisi += parca.isna().any(axis=1).sum()

# Eksik değerlerin yüzdesini hesaplama
eksik_ozet = pd.DataFrame({
    "Eksik Değer Sayısı": eksik_sayilari,
    "Eksik Oranı (%)": (
        eksik_sayilari / toplam_kayit * 100
    ).round(2)
})

# Eksik değeri bulunan sütunları sıralama
eksik_ozet = eksik_ozet[
    eksik_ozet["Eksik Değer Sayısı"] > 0
].sort_values("Eksik Değer Sayısı", ascending=False)

# Analiz sonuçlarını gösterme
print("Toplam kayıt sayısı:", toplam_kayit)
print("Toplam sütun sayısı:", len(eksik_sayilari))
print("Eksik değer içeren satır sayısı:", eksik_satir_sayisi)
print("Eksik değer bulunan sütun sayısı:", len(eksik_ozet))

print("\nEksik Değer Analizi Sonuçları:")
display(eksik_ozet.head(20))

if eksik_ozet.empty:
    print("Standart eksik değer kontrolünde eksik veri bulunamadı.")

```
