### 2.4. Veri Tiplerinin İncelenmesi (Mixed Data Types)

**Amaç:** Edge-IIoTset veri setinde bulunan sayısal, kategorik, metinsel ve ikili değişkenlerin incelenmesi ve veri tiplerinin belirlenmesi.

Bu aşamada farklı trafik türlerinden örnekler alınarak veri setinin karma veri yapısı analiz edilmektedir.


```python

import pandas as pd

# Farklı trafik türlerinden örnek dosyalar seçme
ornek_dosyalar = [
    "DDoS_UDP_Flood_attack.csv",
    "MITM_attack.csv",
    "Distance.csv"
]

# Veri tiplerini karşılaştırmak için sonuç listesi
tip_sonuclari = []

for dosya_adi in ornek_dosyalar:

    # Dosyayı bulma
    dosya = next(
        d for d in ham_dosyalar
        if d.name == dosya_adi
    )

    # Belleği korumak için ilk 1000 kaydı okuma
    ornek_veri = pd.read_csv(dosya, nrows=1000)

    # Veri tiplerini sayma
    tip_sayilari = ornek_veri.dtypes.value_counts()

    print("\nDosya:", dosya_adi)
    print("İncelenen kayıt:", len(ornek_veri))
    print("Sütun sayısı:", len(ornek_veri.columns))

    print("\nVeri tiplerinin dağılımı:")
    print(tip_sayilari)

    # Sonuçları kaydetme
    for veri_tipi, sayi in tip_sayilari.items():
        tip_sonuclari.append({
            "Dosya Adı": dosya_adi,
            "Veri Tipi": str(veri_tipi),
            "Sütun Sayısı": sayi
        })

    # Önemli sütunlardan örnekler gösterme
    onemli_sutunlar = [
        "frame.time",
        "ip.src_host",
        "tcp.dstport",
        "http.request.method",
        "Attack_label",
        "Attack_type"
    ]

    mevcut_sutunlar = [
        s for s in onemli_sutunlar
        if s in ornek_veri.columns
    ]

    print("\nÖrnek veri kayıtları:")
    display(ornek_veri[mevcut_sutunlar].head(3))

# Veri tipi karşılaştırma tablosu
tip_ozeti = pd.DataFrame(tip_sonuclari)

print("\nGenel Veri Tipi Karşılaştırması:")
display(tip_ozeti)
```

### 2.5. Veri Tipi Tutarlılık Analizi 

Edge-IIoTset veri setinde bulunan 24 orijinal CSV dosyasının her birinden ilk 1.000 kayıt incelenerek sütunların veri tipleri karşılaştırılmıştır.


```python

import pandas as pd

# Tüm orijinal CSV dosyalarının veri tiplerini karşılaştırma
veri_tipi_sonuclari = []

for dosya in ham_dosyalar:

    # Her dosyadan ilk 1000 kaydı okuma
    ornek = pd.read_csv(dosya, nrows=1000)

    # Sütunların veri tiplerini kaydetme
    for sutun, veri_tipi in ornek.dtypes.items():
        veri_tipi_sonuclari.append({
            "Dosya Adı": dosya.name,
            "Sütun Adı": sutun,
            "Veri Tipi": str(veri_tipi)
        })

# Sonuçları tabloya dönüştürme
tip_tablosu = pd.DataFrame(veri_tipi_sonuclari)

# Farklı dosyalarda farklı tipte olan sütunları bulma
tip_farklari = (
    tip_tablosu.groupby("Sütun Adı")["Veri Tipi"]
    .agg(lambda x: ", ".join(sorted(x.unique())))
    .reset_index()
)

# Birden fazla veri tipine sahip sütunları seçme
farkli_sutunlar = tip_farklari[
    tip_farklari["Veri Tipi"].str.contains(",")
]

print("İncelenen dosya sayısı:", len(ham_dosyalar))
print("Farklı veri tiplerine sahip sütun sayısı:",
      len(farkli_sutunlar))

print("\nVeri tipi farklılıkları:")
display(farkli_sutunlar)
```

Edge-IIoTset veri setindeki farklı CSV dosyalarının aynı sütunlarda tutarlı veri tipleri içerip içermediğini belirlemek amacıyla veri tipi karşılaştırma analizi gerçekleştirilmiştir.

#### Bulguların Değerlendirilmesi

Analiz sonucunda 63 sütunun 34 tanesinde farklı CSV dosyaları arasında veri tipi farklılıkları bulunduğu belirlenmiştir.

Bu durum, veri setinin sütun isimleri ve sıralamaları bakımından aynı yapıya sahip olmasına rağmen sütun içeriklerinin veri tipi açısından farklılık gösterebildiğini ortaya koymaktadır.

Örneğin `http.request.method` sütunu bazı dosyalarda sayısal, bazı dosyalarda ise metinsel biçimde okunmaktadır.

Bu farklılıkların olası nedenlerinden biri, belirli ağ protokollerine ait olmayan alanlarda sıfır gibi değerlerin kullanılmasıdır.

Pandas kütüphanesi sütunların veri tiplerini mevcut değerlere göre otomatik olarak belirlediğinden, yalnızca sayısal değer içeren bir sütun sayısal, metinsel değerler içeren aynı sütun ise metinsel olarak okunabilmektedir.

Bu nedenle farklı veri tiplerinin tespit edilmesi, ilgili sütunlarda mutlaka hatalı veri bulunduğu anlamına gelmemektedir.

#### Veri Ön İşleme Açısından Önemi

Veri tipi farklılıkları, veri bütünleştirme ve makine öğrenmesi aşamalarında dikkate alınması gereken önemli bir konudur.

Bu kapsamda aşağıdaki işlemlerin değerlendirilmesi planlanmaktadır:

1. Sütunların gerçek içeriklerinin incelenmesi.
2. Sayısal ve metinsel değerlerin anlamlarının belirlenmesi.
3. Sıfır değerlerin eksik veri mi yoksa protokole özgü bir gösterim mi olduğunun araştırılması.
4. Gerekli sütunlara uygun veri tipi dönüşümlerinin uygulanması.
5. Kategorik değişkenler için uygun kodlama yöntemlerinin belirlenmesi.

#### Sonuç

Yapılan analiz, Edge-IIoTset veri setinde dosyalar arasında veri tipi tutarlılığının ayrıca kontrol edilmesi gerektiğini göstermiştir.

Toplam 34 sütunda farklı veri tipi çıkarımları gözlemlenmiş olması, veri temizleme ve öznitelik mühendisliği aşamalarının önemini ortaya koymaktadır.

Ancak bu analiz her dosyanın ilk 1.000 kaydı üzerinden gerçekleştirilmiştir. Bu nedenle sonuçlar tüm kayıtların kesin veri tipi dağılımını temsil etmemektedir.

Mevcut aşamada herhangi bir veri dönüştürme veya silme işlemi uygulanmamış, yalnızca veri kalitesine ilişkin ön bulgular elde edilmiştir.


```python

import pandas as pd

# İncelenecek sütunları belirleme
incelenecek_sutunlar = [
    "frame.time",
    "ip.dst_host",
    "http.request.method",
    "mqtt.topic"
]

# Farklı trafik türlerinden dosya seçme
ornek_dosyalar = [
    "DDoS_UDP_Flood_attack.csv",
    "MITM_attack.csv",
    "Distance.csv"
]

# Analiz sonuçlarını saklama
icerik_sonuclari = []

for dosya_adi in ornek_dosyalar:

    dosya = next(
        d for d in ham_dosyalar
        if d.name == dosya_adi
    )

    # Verileri metin olarak okuyarak orijinal biçimi koruma
    ornek = pd.read_csv(
        dosya,
        usecols=incelenecek_sutunlar,
        nrows=2000,
        dtype=str,
        keep_default_na=False
    )

    for sutun in incelenecek_sutunlar:

        degerler = ornek[sutun].str.strip()

        # Sıfır ve boş görünen değerleri ayırma
        sifir_bos = degerler.isin(["", "0", "0.0"])

        # Diğer değerlerden örnekler seçme
        ornek_degerler = (
            degerler[~sifir_bos]
            .drop_duplicates()
            .head(3)
            .tolist()
        )

        icerik_sonuclari.append({
            "Dosya": dosya_adi,
            "Sütun": sutun,
            "Sıfır/Boş Sayısı": int(sifir_bos.sum()),
            "Diğer Değer Örnekleri": str(ornek_degerler)
        })

# Analiz sonuçlarını gösterme
icerik_tablosu = pd.DataFrame(icerik_sonuclari)

print("Veri Tipi İçerik Analizi:")
display(icerik_tablosu)
```

```python

import pandas as pd
import re

# Farklı zaman biçimlerini incelemek için dosyalar
kontrol_dosyalari = [
    "DDoS_UDP_Flood_attack.csv",
    "MITM_attack.csv",
    "Distance.csv"
]

# Veri biçimini belirleyen fonksiyon
def zaman_bicimini_belirle(deger):
    deger = str(deger).strip()

    if deger == "":
        return "Boş"

    if re.fullmatch(r"\d{1,3}(\.\d{1,3}){3}", deger):
        return "IP biçimine benzer"

    if re.match(r"^\d{4}[-/ ]\d{2}", deger):
        return "Tarih-saat benzeri"

    if re.fullmatch(r"[+-]?\d+(\.\d+)?", deger):
        return "Sayısal"

    return "Diğer"

# Sonuçların saklanması
zaman_sonuclari = []

for dosya_adi in kontrol_dosyalari:

    dosya = next(
        d for d in ham_dosyalar
        if d.name == dosya_adi
    )

    # İlk 2000 kaydı metin olarak okuma
    ornek = pd.read_csv(
        dosya,
        usecols=["frame.time"],
        nrows=2000,
        dtype=str,
        keep_default_na=False
    )

    # Veri biçimlerinin dağılımını hesaplama
    bicimler = ornek["frame.time"].apply(
        zaman_bicimini_belirle
    )

    for bicim, sayi in bicimler.value_counts().items():
        zaman_sonuclari.append({
            "Dosya": dosya_adi,
            "Veri Biçimi": bicim,
            "Kayıt Sayısı": sayi
        })

# Sonuç tablosunu gösterme
zaman_tablosu = pd.DataFrame(zaman_sonuclari)

print("frame.time Biçim Analizi:")
display(zaman_tablosu)
```
